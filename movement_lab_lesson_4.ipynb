```python
import numpy as np

start = np.array([0, 0, 0])     # shoulder, elbow, wrist — arm at rest
end   = np.array([90, 60, 30])  # target pose

motion = np.linspace(start, end, 4)   # whole arm, start → end, over 4 frames
print(motion)
```
