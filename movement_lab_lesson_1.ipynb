```python
import numpy as np

right_arm = np.array([90, 45, 10])   # shoulder, elbow, wrist (degrees)
print(right_arm[2])
print(right_arm * 2)
```

```python
import numpy as np

# two frames of motion — the arm changing over time
# each ROW is one full pose: [shoulder, elbow, wrist]
motion = np.array([
    [90, 45, 10],    # frame 0
    [20, 30, 5]      # frame 1
])

print(motion)
print(motion[0])
```
