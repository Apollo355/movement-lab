```python
import numpy as np

# four frames of motion — each ROW is a full pose [shoulder, elbow, wrist]
motion = np.array([
    [90, 45, 10],    # frame 0
    [70, 40, 12],    # frame 1
    [50, 35, 14],    # frame 2
    [20, 30, 16]     # frame 3
])

wrist = motion[:, 2]        # one joint across every frame — the wrist trajectory
shoulder = motion[:, 0]     # the shoulder trajectory

print("wrist trajectory:  ", wrist)
print("wrist velocity:    ", np.diff(wrist))      # [2 2 2]  → constant = steady
print("shoulder velocity: ", np.diff(shoulder))   # [-20 -20 -30] → changing = accelerates
```
